```python
#experiment 1
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
import re
import unicodedata

# Ensure you have the necessary NLTK data files
nltk.download('punkt')
nltk.download('stopwords')

def mytokenizer(corpus):
    return word_tokenize(corpus)

def text_filter(tokens):
    # Remove punctuation and numbers
    clean_tokens = [re.sub(r'[^\w\s]', '', w) for w in tokens]
    clean_tokens = [re.sub(r'\d+', '', w) for w in clean_tokens]
    return clean_tokens

def remove_stopwords(tokens):
    stop_words = set(stopwords.words('english'))
    return [w for w in tokens if w.lower() not in stop_words]

def my_stemmer(tokens):
    stemmer = PorterStemmer()
    return [stemmer.stem(word) for word in tokens]

def validate_script(text):
    for char in text:
        script = unicodedata.name(char)
        if script != 'Common':  # Exclude common characters like spaces, punctuation, etc.
            return script
    return None  # If no significant characters were found

sentence = ("Pointing to the close123 123 #@relations developed with India by the Awami "
            "League government, Bangladesh Information Minister Hasan Mahmud pointed out that the "
            "country's ties with its larger neighbour have always been affected and minorities have faced "
            "atrocities whenever the opposition BNP comes to power.")

tokens = mytokenizer(sentence)
filtered_tokens = text_filter(tokens)
new_filtered_tokens = remove_stopwords(filtered_tokens)
stem_words = my_stemmer(new_filtered_tokens)

print("The tokens are")
print(tokens)
print("Filtered tokens:")
print(filtered_tokens)
print("The tokens without stop words:")
print(new_filtered_tokens)
print("Stemmed Words")
print(stem_words)

text = "नमस्ते"  # Example text in Devanagari script (Hindi)
script = validate_script(text)
if script:
    print(f"The text is in {script} script.")
else:
    print("Unable to determine the script.")
```

```python
import nltk
from nltk.stem import WordNetLemmatizer

# Ensure you have the necessary NLTK data files
nltk.download('wordnet')

def find_suffixes(word):
    lemmatizer = WordNetLemmatizer()
    suffixes = []
    for pos in ['n', 'v', 'a', 'r']:  # noun, verb, adjective, adverb
        lemma = lemmatizer.lemmatize(word, pos)
        if lemma != word:
            suffix = word[len(lemma):]
            suffixes.append(suffix)
    return suffixes

# Example word
word = "running"
suffixes = find_suffixes(word)
print(f"The suffixes of the word '{word}' are: {suffixes}")
```

```python
#experiment 3
from collections import defaultdict
import numpy as np

class NgramModel:
    def __init__(self, n, k=1):
        self.n = n
        self.k = k
        self.ngram_counts = defaultdict(int)
        self.context_counts = defaultdict(int)

    def train(self, corpus):
        for sentence in corpus:
            words = sentence.split()
            for i in range(len(words) - self.n + 1):
                ngram = tuple(words[i:i+self.n-1])
                next_word = words[i+self.n-1]
                self.ngram_counts[(ngram, next_word)] += 1
                self.context_counts[ngram] += 1

    def probability(self, ngram, next_word):
        context = ngram[:-1]
        context_count = self.context_counts[context]
        ngram_count = self.ngram_counts[(ngram, next_word)]
        return (ngram_count + self.k) / (context_count + self.k * len(self.ngram_counts))

    def generate_next_word(self, ngram):
        possible_next_words = [word for ngram_, word in self.ngram_counts.keys() if ngram_ == ngram]
        probabilities = [self.probability(ngram, word) for word in possible_next_words]
        return np.random.choice(possible_next_words, p=probabilities)

# Example usage
corpus = ["this is a cat", "this is a dog", "this is a cat"]
ngram_model = NgramModel(n=2, k=1)
ngram_model.train(corpus)

ngram = ("is",)
next_word = ngram_model.generate_next_word(ngram)
print(next_word)
```

```python
import nltk
from collections import defaultdict, Counter
from nltk.util import ngrams

# Ensure you have the necessary NLTK data files
nltk.download('punkt')

def calculate_ngram_probabilities(corpus, n):
    ngram_counts = Counter()
    context_counts = Counter()

    for sentence in corpus:
        words = nltk.word_tokenize(sentence)
        n_grams = list(ngrams(words, n))
        for ngram in n_grams:
            context = ngram[:-1]
            ngram_counts[ngram] += 1
            context_counts[context] += 1

    ngram_probabilities = {ngram: (count / context_counts[ngram[:-1]]) for ngram, count in ngram_counts.items()}
    return ngram_probabilities

# Example usage
corpus = ["this is a cat", "this is a dog", "this is a cat"]
n = 2
ngram_probabilities = calculate_ngram_probabilities(corpus, n)

print("N-gram Probabilities:")
for ngram, prob in ngram_probabilities.items():
    print(f"{ngram}: {prob:.4f}")
```

```python
#exercise 4
import nltk

nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')

def find_verbs_and_nouns(sentence):
    words = nltk.word_tokenize(sentence)
    pos_tags = nltk.pos_tag(words)
    
    verbs = [word for word, pos in pos_tags if pos.startswith('VB')]
    nouns = [word for word, pos in pos_tags if pos.startswith('NN')]
    
    return verbs, nouns

# Example sentence
sentence = "The quick brown fox jumps over the lazy dog."

verbs, nouns = find_verbs_and_nouns(sentence)
print(f"Verbs: {verbs}")
print(f"Nouns: {nouns}")
```

```python
#experiment5
from nltk.tokenize import word_tokenize 
from nltk.tag import pos_tag 
from nltk.chunk import RegexpParser 
 
# Input sentence 
sentence = "The quick brown fox jumps over the lazy dog." 
 
# Tokenize the sentence 
tokens = word_tokenize(sentence) 
 
# Perform POS tagging 
pos_tags = pos_tag(tokens) 
 
# Define chunking patterns using regular expressions 
chunking_grammar = r""" 
    NP: {<DT|JJ|NN.*>+} 
    VP: {<VB.*><NP|PP>*} 
""" 
 
# Create a chunk parser 
chunk_parser = RegexpParser(chunking_grammar) 
 
# Parse the POS-tagged text to extract chunks 
chunks = chunk_parser.parse(pos_tags) 
 
# Print the chunks 
chunks.pretty_print() 
```
